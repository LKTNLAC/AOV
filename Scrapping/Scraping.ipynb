```python
page_link = 'https://lienquan.garena.vn/hoc-vien/tuong-skin/'
page_response = requests.get(page_link)
page_content = BeautifulSoup(page_response.content, features='html.parser')
```

```python
tag_hero_list = page_content.find_all(name='a', attrs={'class': 'st-heroes__item'})
print(len(tag_hero_list))
```

```python
list_hero_names = []
list_images = []

for tag_hero in tag_hero_list:
    # Tên tướng
    hero_name = tag_hero.h3.text
    list_hero_names.append(hero_name)

    

    # Hình ảnh
    # Lấy danh sách tên hình đưa vào list_images
    # Download tất cả hình vào thư mục images/man-hinh-lcd/<ten-hinh>
    import os

    # Hình ảnh
    img_tag = tag_hero.find("img")

    # Lấy tên hình
    img_name = img_tag["src"].split("/")[-1]
    list_images.append(img_name)

    # Link hình
    img_url = img_tag["src"]
    if img_url.startswith("//"):
        img_url = "https:" + img_url
    elif img_url.startswith("/"):
        img_url = "https://lienquan.garena.vn" + img_url

    # Tạo thư mục nếu chưa có
    os.makedirs("images/heroes", exist_ok=True)

    # Download hình
    download_image(img_url, f"images/heroes/{img_name}")

print(len(list_hero_names), list_hero_names)
```

```python
import os
import requests
from bs4 import BeautifulSoup

os.makedirs("Images", exist_ok=True)

heroes = soup.select("a.st-heroes__item")

print("Số tướng:", len(heroes))

headers = {
    "User-Agent": "Mozilla/5.0"
}

for hero in heroes:

    # tên tướng
    name = hero.select_one("h2.st-heroes__item--name").get_text(strip=True)

    # ảnh
    img = hero.select_one("img")

    img_url = img["src"]

    print(name, img_url)

    r = requests.get(img_url, headers=headers)

    # giữ đúng đuôi file
    ext = img_url.split(".")[-1].split("?")[0]

    filename = f"Images/{name}.{ext}"

    with open(filename, "wb") as f:
        f.write(r.content)
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(options=options)

driver.get("https://lienquan.garena.vn/hoc-vien/tuong-skin/")

time.sleep(5)

soup = BeautifulSoup(driver.page_source, "html.parser")

heroes = soup.select("a.st-heroes__item")

print(len(heroes))
```

```python
hero_list = []

for hero in heroes:

    name = hero.select_one(".st-heroes__item--name").text.strip()

    detail = hero["href"]

    hero_list.append({
        "name": name,
        "detail": detail
    })

print(hero_list[:5])
```

```python
import requests

headers = {
    "User-Agent":"Mozilla/5.0"
}

for hero in hero_list[:10]:

    html = requests.get(hero["detail"],headers=headers).text

    soup = BeautifulSoup(html,"html.parser")

    print(hero["name"])

    print(soup.title.text)
```

```python
imgs = soup.find_all("img")

print(len(imgs))

for img in imgs:
    print(img.get("src"))
```
