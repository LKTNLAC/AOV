```python
page_link = 'https://lienquan.garena.vn/hoc-vien/tuong-skin/'
page_response = requests.get(page_link)
page_content = BeautifulSoup(page_response.content, features='html.parser')
```

```python
tag_hero_list = page_content.find_all(name='a', attrs={'class': 'st-heroes__item'})
print(len(tag_hero_list))
```

```python
list_hero_names = []
list_images = []

for tag_hero in tag_hero_list:
    # Tên tướng
    hero_name = tag_hero.h3.text
    list_hero_names.append(hero_name)

    

    # Hình ảnh
    # Lấy danh sách tên hình đưa vào list_images
    # Download tất cả hình vào thư mục images/man-hinh-lcd/<ten-hinh>
    import os

    # Hình ảnh
    img_tag = tag_hero.find("img")

    # Lấy tên hình
    img_name = img_tag["src"].split("/")[-1]
    list_images.append(img_name)

    # Link hình
    img_url = img_tag["src"]
    if img_url.startswith("//"):
        img_url = "https:" + img_url
    elif img_url.startswith("/"):
        img_url = "https://lienquan.garena.vn" + img_url

    # Tạo thư mục nếu chưa có
    os.makedirs("images/heroes", exist_ok=True)

    # Download hình
    download_image(img_url, f"images/heroes/{img_name}")

print(len(list_hero_names), list_hero_names)
```

```python
import os
import time
import requests

from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup

# ==========================
# Khởi tạo Chrome
# ==========================
driver = webdriver.Chrome()

url = "https://lienquan.garena.vn/hoc-vien/tuong-skin/"
driver.get(url)

# đợi javascript load
time.sleep(5)

soup = BeautifulSoup(driver.page_source, "html.parser")

driver.quit()

# ==========================
# Tạo thư mục Images
# ==========================
os.makedirs("Images", exist_ok=True)

heroes = []

# Lấy tất cả thẻ img
imgs = soup.find_all("img")

for img in imgs:

    src = (
        img.get("src")
        or img.get("data-src")
        or img.get("data-original")
    )

    alt = img.get("alt")

    if not src or not alt:
        continue

    # chỉ lấy ảnh thật
    if "http" not in src:
        continue

    heroes.append((alt, src))

print("Số tướng:", len(heroes))

# ==========================
# Download ảnh
# ==========================
for name, img_url in heroes:

    filename = f"Images/{name}.jpg"

    try:
        #r = requests.get(img_url)
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/138.0 Safari/537.36",
            "Referer": "https://lienquan.garena.vn/"
        }

        r = requests.get(img_url, headers=headers)
        print(img_url)
        with open(filename, "wb") as f:
            f.write(r.content)

        #print("Downloaded:", name)
        print(r.status_code)
        print(r.url)
        print(r.headers.get("Content-Type"))
        print(len(r.content))
        print(r.text[:200])

    except Exception as e:
        print(name, e)
print(img_url)
```
